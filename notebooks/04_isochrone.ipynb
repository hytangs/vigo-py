# Accessibility surface

Run the native walk-transit-walk accessibility operator from one origin. The result contains the full raster receipt and edge-supported GeoJSON contours for a map or downstream analysis.

```python
import os
from pathlib import Path

import vigo_router

data_home = Path(os.environ.get("VIGO_DATA_HOME", Path.home() / "Documents" / "vigo-data")).expanduser()
origin = [float(value) for value in os.environ.get("VIGO_ORIGIN", "-73.9857,40.7484").split(",")]
with vigo_router.open_network(
    os.environ.get("VIGO_ROUTING_STORE", data_home / "network" / "routing" / "project.sqlite"),
    street_store=os.environ.get("VIGO_STREET_STORE", data_home / "network" / "osm" / "street-index.sqlite"),
) as network:
    surface = network.isochrone(
        origin, cutoffs_minutes=(15, 30, 45), time="08:00",
        service_date=os.environ["VIGO_SERVICE_DATE"],
        service_day=os.environ["VIGO_SERVICE_DAY"], max_walk_km=2.0,
    )

surface.query, len(surface.to_geojson().get("features", []))
```
