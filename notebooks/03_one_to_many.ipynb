# One origin to many destinations

Use one resident native forward search for several destinations. The returned rows keep labels, statuses, route diagnostics, and the exact source policy used by the query.

```python
import os
from pathlib import Path

import vigo_router

data_home = Path(os.environ.get("VIGO_DATA_HOME", Path.home() / "Documents" / "vigo-data")).expanduser()
origin = [float(value) for value in os.environ.get("VIGO_ORIGIN", "-73.9857,40.7484").split(",")]
destinations = {
    "destination-a": [-73.9772, 40.7527],
    "destination-b": [-73.9815, 40.7580],
}
with vigo_router.open_network(
    os.environ.get("VIGO_ROUTING_STORE", data_home / "network" / "routing" / "project.sqlite"),
    street_store=os.environ.get("VIGO_STREET_STORE", data_home / "network" / "osm" / "street-index.sqlite"),
) as network:
    result = network.route_many(
        origin, destinations, time="08:00",
        service_date=os.environ["VIGO_SERVICE_DATE"],
        service_day=os.environ["VIGO_SERVICE_DAY"], max_walk_km=1.2,
    )

[(row.destination_id, row.status, row.duration_minutes) for row in result]
```
