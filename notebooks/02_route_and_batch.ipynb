# Route one journey and a batch

Open the stores built from the same GTFS and OSM source pair, route two coordinates, and retain the full plan plus the batch timing receipt.

```python
import os
from pathlib import Path

import vigo_router

data_home = Path(os.environ.get("VIGO_DATA_HOME", Path.home() / "Documents" / "vigo-data")).expanduser()
service_date = os.environ["VIGO_SERVICE_DATE"]
service_day = os.environ["VIGO_SERVICE_DAY"]
origin = [float(value) for value in os.environ.get("VIGO_ORIGIN", "-73.9857,40.7484").split(",")]
destination = [float(value) for value in os.environ.get("VIGO_DESTINATION", "-73.9772,40.7527").split(",")]
```

```python
with vigo_router.open_network(
    os.environ.get("VIGO_ROUTING_STORE", data_home / "network" / "routing" / "project.sqlite"),
    street_store=os.environ.get("VIGO_STREET_STORE", data_home / "network" / "osm" / "street-index.sqlite"),
) as network:
    plan = network.route(
        origin, destination, time="08:00", service_date=service_date,
        service_day=service_day, max_walk_km=1.2,
    )
    batch = network.route_batch(
        [{"id": "sample", "origin": origin, "destination": destination}],
        service_date=service_date, service_day=service_day, time="08:00",
    )

(plan.status, plan.duration_minutes), (batch.timing, batch[0].status)
```
