# Build a VIGO transport network

Compile one matching GTFS ZIP and OSM PBF into the SQLite stores used by the desktop app, CLI, and Python binding. Keep source inputs and generated artifacts outside the repository.

```python
import os
from pathlib import Path

import vigo_router

data_home = Path(os.environ.get("VIGO_DATA_HOME", Path.home() / "Documents" / "vigo-data")).expanduser()
osm_pbf = Path(os.environ.get("VIGO_OSM", data_home / "input" / "network.osm.pbf")).expanduser()
gtfs_zip = Path(os.environ.get("VIGO_GTFS", data_home / "input" / "network.gtfs.zip")).expanduser()
if not osm_pbf.is_file() or not gtfs_zip.is_file():
    raise FileNotFoundError("Set VIGO_OSM and VIGO_GTFS to matching local inputs.")

with vigo_router.TransportNetwork(
    osm_pbf, [gtfs_zip], cache_dir=data_home / "cache", progress=True
) as network:
    build_summary = network.build_summary

build_summary
```
